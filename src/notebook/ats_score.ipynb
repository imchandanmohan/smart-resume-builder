```python
import re

def extract_metrics(text):
    patterns = [
        r"\$?\d{1,3}(?:,\d{3})+(?:\.\d+)?",                        # $1,000.00 or 22,000
        r"\$?\d+(?:\.\d+)? ?(?:million|billion|k|m|K|M)",          # 2.5 million, $4M, 100K
        r"\$?\d+(?:\.\d+)?%?(?:\+| percent|%)?",                   # 25%, 100, 15% YTD
        r"\d+(?:\.\d+)? ?(?:per week|per month|per day)",         # 10 hours per week
        r"\d+(?:\.\d+)? ?(?:students|firms|people|users|accounts|clients)",  # 100 students
        r"\d+(?:\.\d+)?\+?",                                       # 10+, 100+, 4.5
        r"\d+ ?(?:hours|days|weeks|months)",                      # 10 hours, 3 days
        r"\d+(?:\.\d+)?x",                                         # 3x, 1.5x
        r"\d+ ?gpa",                                               # 4.0 GPA
    ]

    combined_pattern = "|".join(patterns)
    matches = re.findall(combined_pattern, text.lower())
    return [match.strip() for match in matches if match.strip()]

# Example Usage
if __name__ == "__main__":
    sample_text = """
    Increased revenue by $1,200 and improved performance by 3.00%. Supported 10+ accounting firms.
    Worked 10 hours per week for 22,000+ students. Achieved 2.0 AWC and improved retention by 15% YTD.
    Generated $2450 million in client value. Maintained a 4.5 GPA across semesters.
    """

    extracted = extract_metrics(sample_text)
    print("Extracted Metrics:")
    print(extracted)
```

```python
import re

# Large set of impact verbs
impact_starters = {
    # Base forms
    "increase", "improve", "optimize", "deliver", "boost", "generate", "achieve", "reduce",
    "build", "grow", "streamline", "develop", "implement", "manage", "design", "launch",
    "create", "lead", "coordinate", "drive", "maximize", "establish", "support",

    # Past forms
    "increased", "improved", "optimized", "delivered", "boosted", "generated", "achieved", "reduced",
    "built", "grew", "streamlined", "developed", "implemented", "managed", "designed", "launched",
    "created", "led", "coordinated", "drove", "maximized", "established", "supported",

    # Gerund forms
    "increasing", "improving", "optimizing", "delivering", "boosting", "generating", "achieving", "reducing",
    "building", "growing", "streamlining", "developing", "implementing", "managing", "designing", "launching",
    "creating", "leading", "coordinating", "driving", "maximizing", "establishing", "supporting"
}

# Metric pattern
metric_pattern = r"\$?\d{1,3}(?:,\d{3})*(?:\.\d+)?(?:\+|%)?|\d+(?:\.\d+)?(?:x|%)?"

# Stop words and punctuation
stop_words = {
    "and", "or", "but", "so", "because", "although", "though", "however", "while", "instead", "yet",
    "still", "then", "unless", "since", "as", "if", "like"
}
stop_punctuation = {".", ",", ";", "!", "?"}

def extract_impact_phrases(text):
    tokens = text.replace("\n", " ").split()
    phrases = []
    current_phrase = []
    metric_found = False
    capturing = False

    for i, word in enumerate(tokens):
        clean_word = word.strip(".,;!?").lower()

        # Start capturing if this word is an impact verb
        if not capturing and clean_word in impact_starters:
            capturing = True
            current_phrase = [word]
            metric_found = False
            continue

        if capturing:
            current_phrase.append(word)

            # Check for metric presence
            if re.match(metric_pattern, clean_word):
                metric_found = True

            # Stop at stop word or punctuation
            if clean_word in stop_words or any(p in word for p in stop_punctuation):
                if metric_found:
                    phrases.append(" ".join(current_phrase).strip())
                capturing = False
                current_phrase = []
                metric_found = False

    # Check final phrase
    if capturing and metric_found:
        phrases.append(" ".join(current_phrase).strip())

    return phrases

# Example usage
if __name__ == "__main__":
    resume_text = """
    Improved retention by 30%. Reduced churn 15% across the quarter and increased usage. 
    Delivered $1.2 million in revenue. Generated 22,000+ leads through paid ads. Built dashboards with 4.5 GPA data.
    Supported 10+ clients. Implemented changes that drove 2x improvement.
    """

    result = extract_impact_phrases(resume_text)
    print("Impact Phrases:")
    for phrase in result:
        print("-", phrase)
```

```python
import re
import spacy

# Load spaCy model
nlp = spacy.load("en_core_web_sm")

# Lemma-based impact verbs
impact_starters = {
    "improve", "increase", "optimize", "deliver", "boost", "generate", "achieve",
    "reduce", "build", "grow", "streamline", "develop", "implement", "manage",
    "design", "launch", "create", "lead", "coordinate", "drive", "maximize", "establish", "support", "use", "gain", "raise"
}

# Metric pattern (matches %, $, +, x, etc.)
metric_pattern = r"\$?\d{1,3}(?:,\d{3})*(?:\.\d+)?(?:\+|%)?|\d+(?:\.\d+)?(?:x|%)?"

# Stop words/punctuation
stop_words = {"and", "or", "but", "so", "because", "although", "though", "however", "while", "like", "if", "as"}
stop_punctuation = {".", ",", ";", "!", "?"}

def extract_impact_phrases_with_lemma(text, back_window=8):
    doc = nlp(text.replace("\n", " "))
    phrases = []
    current_phrase = []
    capturing = False
    metric_found = False

    for i, token in enumerate(doc):
        word = token.text
        lemma = token.lemma_.lower()

        # Start capturing normally if lemma is an impact starter
        if not capturing and lemma in impact_starters:
            capturing = True
            current_phrase = [word]
            metric_found = False
            continue

        if capturing:
            current_phrase.append(word)
            if re.match(metric_pattern, token.text):
                metric_found = True
            if token.text.lower() in stop_words or token.text in stop_punctuation:
                if metric_found:
                    phrases.append(" ".join(current_phrase).strip())
                capturing = False
                current_phrase = []
                metric_found = False

        # BACKWARD LOOKUP: if a metric appears and we're not already capturing
        if not capturing and re.match(metric_pattern, token.text):
            for j in range(i - 1, max(-1, i - back_window), -1):
                if doc[j].lemma_.lower() in impact_starters:
                    phrase_tokens = doc[j:i+1]
                    phrase = " ".join(tok.text for tok in phrase_tokens).strip()
                    phrases.append(phrase)
                    break

    # Final check
    if capturing and metric_found:
        phrases.append(" ".join(current_phrase).strip())

    return phrases

# Test it
if __name__ == "__main__":
    resume_text = """
    Advised students and faculty during my own office hours for 10 hours per week while implementing impactful projects
such as the Academic TestBank System, ASUCR Career Fair, and various legislation for 22,000+ students.
    """

    results = extract_impact_phrases_with_lemma(resume_text)
    print("Impact Phrases:")
    for phrase in results:
        print("-", phrase)
```
